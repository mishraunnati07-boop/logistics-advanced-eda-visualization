```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set seed for reproducible results
np.random.seed(42)
n = 1200

# 1. Dataset generation
modes = ['Road Freight', 'Express Air', 'Rail Freight', 'Maritime']
hubs = [
    'North Hub (Delhi)',
    'West Hub (Mumbai)',
    'South Hub (Bengaluru)',
    'East Hub (Kolkata)',
]

assigned_modes = np.random.choice(modes, size=n, p=[0.45, 0.25, 0.20, 0.10])
origin_hubs = np.random.choice(hubs, size=n, p=[0.35, 0.30, 0.20, 0.15])
distances = np.random.uniform(150, 2400, size=n)
weights = np.random.exponential(scale=1800, size=n) + 50

# Mode speed and transit calculation
speeds = {
    'Road Freight': 52.0,
    'Express Air': 540.0,
    'Rail Freight': 38.0,
    'Maritime': 22.0,
}
base_speed = np.array([speeds[m] for m in assigned_modes])
transit_hours = (distances / base_speed) + np.random.gamma(
    shape=2.5, scale=2.0, size=n
)

# Cost multipliers
rate_km = {
    'Road Freight': 1.45,
    'Express Air': 6.80,
    'Rail Freight': 0.85,
    'Maritime': 0.50,
}
mode_rate = np.array([rate_km[m] for m in assigned_modes])
costs = (
    (distances * mode_rate) + (weights * 0.28) + np.random.normal(120, 35, size=n)
)

df = pd.DataFrame({
    'Shipment_ID': [f'SHP-{i}' for i in range(1001, 1001 + n)],
    'Origin_Hub': origin_hubs,
    'Transport_Mode': assigned_modes,
    'Distance_KM': np.round(distances, 2),
    'Weight_KG': np.round(weights, 2),
    'Transit_Hours': np.round(transit_hours, 2),
    'Total_Cost_USD': np.round(costs, 2),
})

# 2. Key statistical metrics
summary = (
    df.groupby('Transport_Mode')
    .agg(
        Shipments=('Shipment_ID', 'count'),
        Mean_Transit_Hrs=('Transit_Hours', 'mean'),
        Median_Transit_Hrs=('Transit_Hours', 'median'),
        Mean_Cost=('Total_Cost_USD', 'mean'),
        Median_Cost=('Total_Cost_USD', 'median'),
        Cost_IQR=(
            'Total_Cost_USD',
            lambda x: np.percentile(x, 75) - np.percentile(x, 25),
        ),
    )
    .round(2)
    .reset_index()
)

print('--- SUMMARY TABLE ---')
print(summary.to_string(index=False))

# 3. Generating Visualizations
plt.style.use(
    'seaborn-v0_8-whitegrid'
    if 'seaborn-v0_8-whitegrid' in plt.style.available
    else 'default'
)

# Figure 1: Transit Time Distribution
plt.figure(figsize=(8, 4.5))
sns.kdeplot(
    data=df,
    x='Transit_Hours',
    hue='Transport_Mode',
    fill=True,
    common_norm=False,
)
plt.title('Transit Time Distribution by Transport Mode')
plt.xlabel('Transit Duration (Hours)')
plt.ylabel('Density')
plt.tight_layout()
plt.savefig('vis1_transit_distribution.png', dpi=200)
plt.close()

# Figure 2: Correlation Heatmap
plt.figure(figsize=(6, 4.5))
sns.heatmap(
    df[
        ['Distance_KM', 'Weight_KG', 'Transit_Hours', 'Total_Cost_USD']
    ].corr(),
    annot=True,
    cmap='Blues',
    fmt='.2f',
)
plt.title('Correlation Matrix of Key Metrics')
plt.tight_layout()
plt.savefig('vis2_correlation_matrix.png', dpi=200)
plt.close()

# Figure 3: Cost vs. Weight by Mode
plt.figure(figsize=(8, 4.5))
sns.scatterplot(
    data=df,
    x='Weight_KG',
    y='Total_Cost_USD',
    hue='Transport_Mode',
    alpha=0.7,
)
plt.title('Transportation Cost vs. Consignment Weight')
plt.xlabel('Weight (KG)')
plt.ylabel('Cost ($ USD)')
plt.tight_layout()
plt.savefig('vis3_cost_vs_weight.png', dpi=200)
plt.close()

# Figure 4: Hub Delays & Outliers
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df, x='Origin_Hub', y='Transit_Hours', palette='muted')
plt.title('Transit Time Variation Across Regional Origin Hubs')
plt.xlabel('Dispatch Hub')
plt.ylabel('Transit Time (Hours)')
plt.tight_layout()
plt.savefig('vis4_hub_delays.png', dpi=200)
plt.close()

print('\nEDA Completed successfully. All 4 plots saved.')
```
